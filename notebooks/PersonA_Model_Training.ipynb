```python
from google.colab import drive
import os
import pandas as pd

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Automatically locate SHR76_25a.csv in Drive
print("Searching for SHR76_25a.csv in Google Drive...")
dataset_path = None

for root, dirs, files in os.walk('/content/drive/MyDrive'):
    for file in files:
        if 'shr76' in file.lower():
            dataset_path = os.path.join(root, file)
            break

# 3. Load dataset
if dataset_path:
    print(f"Found file at: {dataset_path}")
    df = pd.read_csv(dataset_path, low_memory=False)
    print(f"Dataset successfully loaded! Total rows: {len(df):,}")
    print("Columns:", df.columns.tolist()[:15])
else:
    print("ERROR: File not found. Make sure SHR76_25a.csv is uploaded inside your Google Drive.")
```

```python
# MAP Code Dictionaries
weapon_dict = {
    11: 'Firearm (Handgun)', 12: 'Firearm (Rifle)', 13: 'Firearm (Shotgun)',
    14: 'Firearm (Unspecified)', 15: 'Firearm (Other)', 20: 'Knife/Cutting Instrument',
    30: 'Blunt Object', 40: 'Personal Weapon (Hands/Fists/Feet)', 50: 'Poison',
    60: 'Pushed Out Window', 70: 'Explosives', 80: 'Fire', 85: 'Narcotics/Drugs',
    90: 'Drowning', 91: 'Strangulation/Asphyxiation', 99: 'Unknown Weapon'
}

sex_dict = {1: 'Male', 2: 'Female', 9: 'Unknown'}

circumstance_dict = {
    2: 'Lovers Triangle', 3: 'Child Killed by Babysitter', 10: 'Brawl Under Alcohol',
    18: 'Argument Over Money', 20: 'Other Argument', 40: 'Gangland Killings',
    50: 'Institutional/Felony', 60: 'Robbery', 70: 'Burglary', 80: 'Narcotic Drug Laws',
    99: 'Undetermined/Unknown'
}

relationship_dict = {
    1: 'Spouse/Partner', 2: 'Child/Parent', 3: 'Other Family',
    4: 'Acquaintance/Friend', 5: 'Stranger', 99: 'Unknown Relationship'
}

# Apply dictionary mappings
df['Weapon_Label'] = df['Weapon'].map(weapon_dict).fillna('Unknown Weapon')
df['VicSex_Label'] = df['VicSex'].map(sex_dict).fillna('Unknown')
df['Circumstance_Label'] = df['Circumstance'].map(circumstance_dict).fillna('Undetermined/Unknown')
if 'Relationship' in df.columns:
    df['Relationship_Label'] = df['Relationship'].map(relationship_dict).fillna('Unknown Relationship')

# Clean Victim Age (Set unknown codes 99, 999 to NaN/Median)
df['VicAge_Clean'] = pd.to_numeric(df['VicAge'], errors='coerce')
df['VicAge_Clean'] = df['VicAge_Clean'].apply(lambda x: np.nan if x in [99, 999] or x > 100 else x)
df['VicAge_Clean'] = df['VicAge_Clean'].fillna(df['VicAge_Clean'].median())

print("Categorical variables mapped and numeric fields cleaned.")
```

```python
# Safely convert 'Solved' column text ('Yes'/'No') into binary integers (1 or 0)
if 'Solved' in df.columns:
    solved_str = df['Solved'].astype(str).str.strip().str.lower()
    df['Solved'] = solved_str.apply(lambda x: 1 if x in ['yes', '1', '1.0', 'true', 'solved'] else 0)
elif 'ActionType' in df.columns:
    df['Solved'] = df['ActionType'].apply(lambda x: 0 if str(x).lower() in ['unsolved', '0', 'no'] else 1)

print(f"Solved Cases: {sum(df['Solved'] == 1):,}")
print(f"Unsolved Cases: {sum(df['Solved'] == 0):,}")
```

```python
# 1. Dataset for Feature 4: Solvability Neural Network & Feature 3: BBN
df_full_clean = df[['Solved', 'Weapon_Label', 'VicAge_Clean', 'VicSex_Label', 'Circumstance_Label']].dropna()

# 2. Dataset for Feature 1: Serial Crime Linking (Clustering Unsolved Cold Cases)
df_unsolved = df[df['Solved'] == 0][['Weapon_Label', 'VicAge_Clean', 'VicSex_Label', 'Circumstance_Label']].dropna()

# One-Hot Encode categorical attributes for Scikit-Learn K-Means & Hierarchical Clustering
X_unsolved_encoded = pd.get_dummies(df_unsolved, columns=['Weapon_Label', 'VicSex_Label', 'Circumstance_Label'], drop_first=True)

print(f"Full Clean Dataset Matrix Shape (for Solvability ANN): {df_full_clean.shape}")
print(f"Unsolved Feature Matrix Shape (for Serial Crime Clustering): {X_unsolved_encoded.shape}")
```

```python
import os

output_dir = '/content/drive/MyDrive/Crimora_ML_Project/datasets/processed'
os.makedirs(output_dir, exist_ok=True)

# Save processed DataFrames
df_full_clean.to_csv(f'{output_dir}/shr_full_processed.csv', index=False)
X_unsolved_encoded.to_csv(f'{output_dir}/shr_unsolved_clustering_matrix.csv', index=False)

print("Day 2 Complete! Cleaned matrices saved to Google Drive /datasets/processed/")
```

```python
import pandas as pd
import numpy as np
import joblib
import os
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

# 1. Load the unsolved case matrix from Day 2
data_path = '/content/drive/MyDrive/Crimora_ML_Project/datasets/processed/shr_unsolved_clustering_matrix.csv'
X_unsolved = pd.read_csv(data_path)

# Sample 20,000 cases if dataset is very large for optimal hierarchical memory efficiency
if len(X_unsolved) > 20000:
    X_sample = X_unsolved.sample(n=20000, random_state=42)
else:
    X_sample = X_unsolved.copy()

# 2. Scale features for distance calculations
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_sample)

print(f"Matrix loaded & scaled successfully. Shape: {X_scaled.shape}")
```

```python
# Create models directory in Google Drive
models_dir = '/content/drive/MyDrive/Crimora_ML_Project/models'
os.makedirs(models_dir, exist_ok=True)

# 1. Train K-Means
n_clusters = 5
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

# 2. Save K-Means Model & Scaler artifact
joblib.dump(kmeans, f'{models_dir}/kmeans_serial.pkl')
joblib.dump(scaler, f'{models_dir}/clustering_scaler.pkl')

print(f"K-Means training complete ({n_clusters} clusters). Saved to {models_dir}/kmeans_serial.pkl")
```

```python
# 1. Train Agglomerative Hierarchical Model
hierarchical = AgglomerativeClustering(n_clusters=n_clusters, linkage='ward')
hierarchical_labels = hierarchical.fit_predict(X_scaled)

# 2. Save Hierarchical Model artifact
joblib.dump(hierarchical, f'{models_dir}/hierarchical_serial.pkl')

print(f"Agglomerative Clustering complete. Saved to {models_dir}/hierarchical_serial.pkl")
```

```python
# Assign predicted clusters back to sample dataframe for inspection
X_sample['KMeans_Cluster'] = kmeans_labels
X_sample['Hierarchical_Cluster'] = hierarchical_labels

# Calculate Silhouette Score for K-Means performance check
score = silhouette_score(X_scaled[:5000], kmeans_labels[:5000])
print(f"=== Model Performance Check ===")
print(f"K-Means Silhouette Score: {score:.4f}")
print("\n=== Cases per K-Means Cluster ===")
print(X_sample['KMeans_Cluster'].value_counts().sort_index())
```
